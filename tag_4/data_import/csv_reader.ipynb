```python
import numpy as np

# data = np.loadtxt("measurement_1.csv", delimiter=",", skiprows=1)
# genfromtxt() kann mit fehlenden Daten umgehen, NaN einsetzen, Strings und Zahlen verarbeiten.

data = np.genfromtxt(
    "measurement_1.csv",
    delimiter=',',
    names=True
)

print(type(data))

# print(data)
time = data["time"]
voltage = data["voltage"]
current = data["current"]

# print("Time Vector:")
# print(time)

# print("Voltage Vector:")
# print(voltage)

power = voltage * current

print(power[:5])

output = np.column_stack((time, power))
np.savetxt("t_vs_power.csv", output, header="time,power", comments="", delimiter=",")
```
