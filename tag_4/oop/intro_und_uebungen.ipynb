### Ziel OOP Block

- Daten werden in ein Objekt
- Berechnungen leben als Methoden im Objekt
- Plots werden vom Objekt erzeugt
- Animation werden vom Objekt erzeugt
- numerische Auswertungen sind sauber gekapselt

```python
# Bisher haben wir einzelne Werkzeuge benutzt
# Jetzt bauen wir daraus ein strukturiertes Analysewerkzeug


import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("measurement.csv", delimiter=",", skiprows=1)
time = data[:,0]
voltage = data[:, 1]
current = data[:, 2]

power = voltage * current

plt.plot(time, voltage)
plt.show()
```

```python
# OOP-Version
# Klasse MeasurementData:
# - lädt Daten
# - Berechnet Leistung
# - Berechnet Mittelwerte
# - erstellt Plot
# - erstellt Animation

class MeasurementData:
    def __init__(self, filename):
        self.filename = filename
        self.data = np.loadtxt(filename, delimiter=",", skiprows=1)
        self.time = self.data[:, 0]
        self.voltage = self.data[:, 1]
        self.current = self.data[:, 2]
```

Zur Wiederholung:
- class definiert einen neuen Datentyp
- init ist der Konstruktor
- self bedeutet: dieses konkrete Objekt
- Attribute speichern Daten des Objekts
