```python
"""
Lösung: Analyse von Messdaten
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# =========================
# 1. Daten importieren
# =========================

data = np.genfromtxt("measurement.csv", delimiter=",", skip_header=1)

time = data[:,0]
voltage = data[:,1]
current = data[:,2]

# =========================
# 2. Daten prüfen
# =========================

if np.isnan(data).any():
    print("Warnung: NaN Werte gefunden")

if np.any(voltage > 6):
    print("Warnung: ungewöhnlich hohe Spannung")

# =========================
# 3. Daten bereinigen
# =========================

# NaN entfernen
data = data[~np.isnan(data).any(axis=1)]

time = data[:,0]
voltage = data[:,1]
current = data[:,2]

# z-Score berechnen: wie weit ist dieser Wert vom Mittelwert entfernt?
z = np.abs(stats.zscore(data))

# Ausreißer entfernen
data_clean = data[(z < 3).all(axis=1)]

time = data_clean[:,0]
voltage = data_clean[:,1]
current = data_clean[:,2]

# =========================
# 4. Berechnungen
# =========================

power = voltage * current

print("Mittlere Spannung:", np.mean(voltage))
print("Mittlerer Strom:", np.mean(current))
print("Mittlere Leistung:", np.mean(power))

# =========================
# 5. Plot
# =========================

plt.plot(time, voltage, label="Voltage")
plt.plot(time, power, label="Power")

plt.xlabel("Time (s)")
plt.ylabel("Value")
plt.legend()
plt.grid()

plt.show()
```
