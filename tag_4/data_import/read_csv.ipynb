```python
# Übung 2: Umgang mit fehlenden Werten

"""
Ausgangssituation:
Eine Messdatei enthält fehlende Werte.
Diese sollen trotzdem eingelesen werden.
"""

# 1. Datei mit genfromtxt laden
# 2. Prüfe, welche Werte NaN sind
# 3. Entferne Zeilen mit NaN

import numpy as np

data = np.genfromtxt("measurement_2.csv", delimiter=",", skip_header=1)

# NaN erkennen
print("ist NaN?:", np.isnan(data))


# Zeilen ohne NaN behalten
clean_data = data[~np.isnan(data).any(axis=1)]

print(clean_data)
```

```python
# Übung 3: Spaltennamen vewrwenden mit names=True

"""
Ausgangssituation:
Messdaten sollen über Spaltennamen angesprochen werden.
"""

# 1. Datei mit Namen laden
# 2. Greife auf "voltage" und "current" zu
# 3. Berechne Leistung

# genfromtxt() - kann mit fehlenden Daten umgehen, NaN einsetzen, Strings und Zahlen verarbeiten.
data = np.genfromtxt(
    "measurement_2.csv",
    delimiter=',',
    names=True
)

voltage = data["voltage"]
current = data["current"]

power = voltage * current

print(power[:5])
```
