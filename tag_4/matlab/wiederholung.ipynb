```python
"""
Ausgangssituation:
Ein Kondensator entlädt sich in einem RC-Kreis.
Die Spannung folgt näherungsweise der Formel:

U(t) = U0 * exp(-t / tau)

mit:
U0  = Anfangsspannung
tau = Zeitkonstante

Die Spannung soll berechnet, geplottet und ausgewertet werden.
Zusätzlich soll bestimmt werden, wann die Spannung unter 1 V fällt.
"""
```

```python
# Teilaufgaben
# 1. Erzeuge einen Zeitvektor von 0 bis 10 Sekunden.
# 2. Setze U0 = 5 V und tau = 2 s.
# 3. Berechne die Spannung U(t).
# 4. Stelle die Spannung über der Zeit dar.
# 5. Berechne numerisch die Fläche unter der Kurve von 0 bis 10 s.
# 6. Finde numerisch den Zeitpunkt, an dem U(t) = 1 V.
# 7. Speichere den Plot als PNG.
```

```python
# Entladung eines RC-Kreises soll simuliert und ausgewertet werden

import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate, optimize

# Startbedingungen

# Anganfspannung in Volt
U0 = 5.0

# Zeitkonstante in Sekunden
tau = 2.0

# np.linspace() erzeugt 300 Zeitpunkte von 0 bis 10 Sekunden.
t = np.linspace(0, 10, 300)

# Berechnen der Entladungspannung für alle Zeitpunkte
U = U0 * np.exp(-t / tau)

# Neue Grafik erstellen
plt.figure()    # Fenster erstellen
plt.plot(t, U, label="U(t)") # Plotten

# Beschrifen Achsen und Grafik
plt.xlabel("Zeit [s]")
plt.ylabel("Spannung [V]")
plt.title("RC-Entladung")
plt.legend()
plt.grid()

plt.show()
```

```python
# Entladung eines RC-Kreises soll simuliert und ausgewertet werden

import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate, optimize

# Startbedingungen

# Anganfspannung in Volt
U0 = 5.0

# Zeitkonstante in Sekunden
tau = 2.0

# np.linspace() erzeugt 300 Zeitpunkte von 0 bis 10 Sekunden.
t = np.linspace(0, 10, 300)

# Berechnen der Entladungspannung für alle Zeitpunkte
U = U0 * np.exp(-t / tau)

# Neue Grafik erstellen
plt.figure()    # Fenster erstellen
plt.plot(t, U, label="U(t)") # Plotten

# Beschrifen Achsen und Grafik
plt.xlabel("Zeit [s]")
plt.ylabel("Spannung [V]")
plt.title("RC-Entladung")
plt.legend()
plt.grid()

plt.savefig("rc_entladung.png")

plt.show()

# 5. Berechne numerisch die Fläche unter der Kurve von 0 bis 10 s.
# 6. Finde numerisch den Zeitpunkt, an dem U(t) = 1 V.

# Diese Hilfsfunktion ist Null, wenn die Spannung auch 1V beträgt. 
def target_function(zeit):
    return U0 * np.exp(-zeit / tau) - 1.0

# Diese Funktion wird für Integration und Nullstellensuche benötigt.
def voltage_function(zeit):
    return U0 * np.exp(-zeit / tau)

# integrate.quad() berechnet die Fläche unter U(t) zwischen 0 und 10 Sekunden.

area, error = integrate.quad(voltage_function, 0, 10)

print("Fläche unter der Kurve:", area)
print("Geschätzter Integrationsfehler:", error)

# optimize.fsolve() findet numerisch den Zeitpunkt, an dem U(t) = 1V gilt.
zeit_1V  = optimize.fsolve(target_function, x0=3.0)
print(zeit_1V)
print(type(zeit_1V))
print("Zeitpunkt für U(t) = 1V:", round(zeit_1V[0], 2), "s")
```
