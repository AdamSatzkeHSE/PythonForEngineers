# Rechnen mit Vektoren und Matrizen

```python
"""
Ausgangssituation:
In MATLAB werden für Simulationen oft Vektoren mit linspace erzeugt
und anschließend mathematische Funktionen darauf angewendet.

Erstelle in Python einen Vektor von 0 bis 10 mit 100 Punkten
und berechne daraus das Signal y = sin(x).
Gib die ersten 5 Werte von x und y aus.
"""


import numpy as np

# np.linspace(start, stop, num) erzeugt num gleichmäßig verteilte Werte
# zwischen start und stop.
x = np.linspace(0, 10, 100)

# np.sin(x) berechnet den Sinus für jedes Element im Array x.
y = np.sin(x)

# Ausgabe der ersten 5 Werte zur Kontrolle.
print("x:", x[:5])
print("y:", y[:5])

```

```python
"""
Ausgangssituation:
Für eine einfache technische Berechnung sollen zwei Vektoren erzeugt werden.
Ein Vektor soll nur Einsen enthalten, ein anderer nur Nullen.
Danach soll ein neuer Vektor berechnet werden.
"""

# 1. Erzeuge ein Array mit fünf Einsen
# 2. Erzeuge ein Array mit fünf Nullen
# 3. Addiere zu den Einsen den Wert 3
# 4. Gib das Ergebnis aus
```

```python
"""
Ausgangssituation:
Einfache Array-Operationen sollen wie in MATLAB durchgeführt werden.
"""

import numpy as np

# np.ones(5) erzeugt ein Array mit 5 Einsen.
a = np.ones(5)

# np.zeros(5) erzeugt ein Array mit 5 Nullen.
b = np.zeros(5)

# Addition eines Skalars zu jedem Element von a.
c = a + 3

print("a:", a)
print("b:", b)
print("c:", c)
```

```python
"""
Ausgangssituation:
In der Elektrotechnik und Regelungstechnik werden oft Matrizen verwendet.
Erzeuge zwei Matrizen und berechne ihr Matrixprodukt.
"""
"""
Zwei Matrizen sollen definiert und miteinander multipliziert werden.
"""

import numpy as np

# np.array() erzeugt eine Matrix aus verschachtelten Listen.
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Der Operator @ berechnet das Matrixprodukt.
C = A @ B

print("A =")
print(A)
print("B =")
print(B)
print("C =")
print(C)
```

```python
"""
Ausgangssituation:
Ein lineares Gleichungssystem soll wie in MATLAB numerisch gelöst werden.

2x + y = 5
x + 3y = 6
"""
```

```python
"""
Ausgangssituation:
Ein lineares Gleichungssystem aus zwei Gleichungen soll gelöst werden.
"""

import numpy as np

# Koeffizientenmatrix des Gleichungssystems.
A = np.array([[2, 1],
              [1, 3]])

# Rechte Seite des Gleichungssystems.
b = np.array([5, 6])

# np.linalg.solve(A, b) berechnet die Lösung x des Systems A*x = b.
x = np.linalg.solve(A, b)

print("Lösung:")
print("x =", x[0])
print("y =", x[1])
```

```python
"""
Ausgangssituation:
Ein Signal soll dargestellt werden.
Erzeuge x von 0 bis 2*pi und plotte sin(x) und cos(x) in einem Diagramm.
Beschrifte Achsen und füge eine Legende hinzu.
"""

```

```python
"""
Ausgangssituation:
Zwei mathematische Signale sollen zusammen geplottet werden.
"""

import numpy as np
import matplotlib.pyplot as plt

# Erzeugt 200 Werte zwischen 0 und 2*pi.
x = np.linspace(0, 2 * np.pi, 200)

# Berechnet Sinus und Kosinus elementweise.
y1 = np.sin(x)
y2 = np.cos(x)

# plt.figure() erstellt eine neue Abbildung.
plt.figure()

# plt.plot(x, y, label=...) zeichnet eine Kurve und versieht sie mit einem Namen.
plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")

# Beschriftung der Achsen.
plt.xlabel("x")
plt.ylabel("y")

# Titel der Grafik.
plt.title("Sinus und Kosinus")

# Zeigt eine Legende an.
plt.legend()

# Zeigt ein Gitternetz.
plt.grid()

# Zeigt die fertige Grafik an.
plt.show()
```

```python
"""
Ausgangssituation:
In technischen Berechnungen müssen oft Nullstellen numerisch bestimmt werden.
Bestimme eine Lösung der Gleichung:

x^3 - x - 2 = 0
"""

"""
Eine nichtlineare Gleichung soll numerisch gelöst werden.
"""

from scipy import optimize

# Diese Funktion beschreibt die Gleichung, deren Nullstelle gesucht wird.
def f(x):
    return x**3 - x - 2

# optimize.fsolve(func, x0) sucht eine Nullstelle von func
# und verwendet x0 als Startwert.
root = optimize.fsolve(f, x0=1.5)

print("Gefundene Nullstelle:", root[0])
```

```python
"""
Ausgangssituation:
Die Fläche unter einer Kurve soll numerisch bestimmt werden.
Berechne das Integral von sin(x) im Bereich von 0 bis pi.
"""
```

```python
"""
Ausgangssituation:
Eine Funktion soll numerisch integriert werden.
"""

import numpy as np
from scipy import integrate

# Zu integrierende Funktion.
def f(x):
    return np.sin(x)

# integrate.quad(func, a, b) berechnet das Integral von func im Bereich [a, b].
result, error = integrate.quad(f, 0, np.pi)

print("Integral:", result)
print("Geschätzter Fehler:", error)
```

```python
"""
Ausgangssituation:
Ein Sensorsystem liefert nur wenige Messpunkte.
Zwischenwerte sollen durch Interpolation bestimmt werden.
"""
```

```python
"""
Aus wenigen Messpunkten sollen Zwischenwerte geschätzt werden.
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy import interpolate

# Gemessene Stützstellen.
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 0.5, 1.5, 1.0])

# interpolate.interp1d(x, y, kind="linear") erstellt eine lineare Interpolationsfunktion.
f = interpolate.interp1d(x, y, kind="linear")

# Neue feinere x-Werte für die Darstellung.
x_new = np.linspace(0, 4, 100)

# Berechnet die interpolierten y-Werte.
y_new = f(x_new)

plt.figure()
plt.plot(x, y, "o", label="Messpunkte")
plt.plot(x_new, y_new, label="Interpolation")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Lineare Interpolation")
plt.legend()
plt.grid()
plt.show()
```
