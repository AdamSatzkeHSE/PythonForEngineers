## Animationen mit Matplotlib

Matplotlib kann nicht nur plotten, es kann auch dynamische Prozesse visualisieren.

Geeignet für:
- Signale über Zeit
- Messdaten
- Simulationen
- Regelungssysteme


```python
from matplotlib.animation import FuncAnimation
# FuncAnimation: ruft wiederholt eine Update-Funktion auf und aktualisiert den Plot

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 100)

signal = np.sin(t)

# Leere Linie initialisieren


fig, ax = plt.subplots()
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

line, = ax.plot([],[])

# Gibt einen Tupel mit Linie zurück.
def update(frame):
    # Zeigt Singal bis zum aktuellen Zeitpunkt
    x = t[:frame]
    # aktualisiert die Linie
    line.set_data(x, signal)

    # 
    return line, 


ani = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=50 # Zeit zwischen Frames in ms
)

plt.show()
```

## Zwei Kurven animieren

```python
from matplotlib.animation import FuncAnimation
# FuncAnimation: ruft wiederholt eine Update-Funktion auf und aktualisiert den Plot

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import HTML

t = np.linspace(0, 10, 100)
fig, ax = plt.subplots()
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)


line1, = ax.plot([], [], label="Sinus")
line2, = ax.plot([], [], label="Cosinus")

def update(frame):
    line1.set_data(t[:frame], np.sin(t[:frame]))
    line2.set_data(t[:frame], np.cos(t[:frame]))

    return line1, line2,

ani = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=50 # Zeit zwischen Frames in ms
)

# Animation speichern
plt.close(fig)
ani.save("animation.gif", writer="pillow")
HTML(ani.to_jshtml())
```
