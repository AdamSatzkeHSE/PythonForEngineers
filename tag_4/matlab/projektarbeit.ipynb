"""
Ausgangssituation:
Die Entladung eines RC-Kreises soll simuliert und ausgewertet werden.
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
from scipy import optimize

# Anfangsspannung in Volt.
U0 = 5.0

# Zeitkonstante in Sekunden.
tau = 2.0

# np.linspace() erzeugt 300 Zeitpunkte von 0 bis 10 Sekunden.
t = np.linspace(0, 10, 300)

# Berechnet die Entladungsspannung für alle Zeitpunkte.
U = U0 * np.exp(-t / tau)

# Erstellt eine neue Grafik.
plt.figure()

# Zeichnet den Spannungsverlauf.
plt.plot(t, U, label="U(t)")

# Beschriftet Achsen und Grafik.
plt.xlabel("Zeit (s)")
plt.ylabel("Spannung (V)")
plt.title("RC-Entladung")
plt.legend()
plt.grid()

# Speichert den Plot als PNG-Datei.
plt.savefig("rc_entladung.png")

# Zeigt die Grafik an.
plt.show()

# Diese Funktion wird für Integration und Nullstellensuche benötigt.
def voltage_function(t_value):
    return U0 * np.exp(-t_value / tau)

# integrate.quad() berechnet die Fläche unter U(t) zwischen 0 und 10 Sekunden.
area, error = integrate.quad(voltage_function, 0, 10)

print("Fläche unter der Kurve:", area)
print("Geschätzter Integrationsfehler:", error)

# Diese Hilfsfunktion ist Null, wenn die Spannung genau 1 V beträgt.
def target_function(t_value):
    return U0 * np.exp(-t_value / tau) - 1.0

# optimize.fsolve() findet numerisch den Zeitpunkt, an dem U(t) = 1 V gilt.
time_at_1V = optimize.fsolve(target_function, x0=3.0)

print("Zeitpunkt für U(t) = 1 V:", time_at_1V[0], "s")
